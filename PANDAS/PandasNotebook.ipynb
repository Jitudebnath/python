#series

A pandas series is like a column in a table . It is a 1-D array holding data of any type

```python
countries=['India','Pakistan','USA','Nepal','SriLanka']
pd.Series(countries)
```

```python
#integers
runs=[13,23,45,67]
pd.Series(runs,dtype='Int32')
```

```python
#custom index
marks=[67,57,89,100]
subjects=['maths','english','science','hindi']

costum_index=pd.Series(marks,index=subjects)
print(costum_index)
```

```python
#setting a name
pd.Series(marks,index=subjects,name='Nihar ke marks ')
```

Series from dict

```python
marks={
    'maths':78,
    'english':89,
    'science':90,
    'hindi':99
}

marks_series=pd.Series(marks,name='jitu marks')
print(marks_series)
```

```python
#size
marks_series.size
```

```python
#dtype
marks_series.dtype
```

```python
#is_unique
marks_series.is_unique
```

```python
#index
marks_series.index
```

```python
#values
marks_series.values
```
