#series

A pandas series is like a column in a table . It is a 1-D array holding data of any type

```python
countries=['India','Pakistan','USA','Nepal','SriLanka']
pd.Series(countries)
```

```python
#integers
runs=[13,23,45,67]
pd.Series(runs,dtype='Int32')
```

```python
#custom index
marks=[67,57,89,100]
subjects=['maths','english','science','hindi']

costum_index=pd.Series(marks,index=subjects)
print(costum_index)
```

```python
#setting a name
pd.Series(marks,index=subjects,name='Nihar ke marks ')
```

```python
#copy 
import pandas as pd

# Original Series
runs = pd.Series([13, 23, 45, 67], dtype="int32")

# Create a copy
runs_copy = runs.copy()

# Modify the copy
runs_copy[0] = 100

print("Original Series:")
print(runs)

print("\nCopied Series:")
print(runs_copy)
```
