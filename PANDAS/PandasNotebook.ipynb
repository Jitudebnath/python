```python
#apply (series and dataframe) it works row wise
points_df=pd.DataFrame({
    '1st point':[(3,4),(4,5),(0,0),(6,8),(4,4),(5,5)],
    '2nd point':[(1,2),(2,3),(0,6),(4,5),(2,3),(3,1)]
})

points_df
```

```python
def eucliden(row):
    pt_A=row['1st point']
    pt_B=row['2nd point']

    return ((pt_A[0]-pt_B[1])**2+(pt_A[1]-pt_B[1])**2)**0.5
```

```python
points_df.apply(eucliden,axis=1)
```

```python
points_df['distance']=points_df.apply(eucliden,axis=1)
points_df
```
