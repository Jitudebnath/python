```python
#can we have multiple index ?
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
a = pd.Series([1,2,3,4,5,6,7,8],index=index_val)
a
```

```python
#multiindexing or hierarchial indexing
#multiple index levels within a single box 

# how to create multiindex object
# 1. pd.MultiIndex.from_tuples()
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
multiindex = pd.MultiIndex.from_tuples(index_val)
multiindex.levels[1]
# 2. pd.MultiIndex.from_product()
pd.MultiIndex.from_product([['cse','ece'],[2019,2020,2021,2022]])
```

```python
# level inside multiindex object
s=pd.Series([1,2,3,4,5,6,7,8],index=multiindex)
s
```

```python
#how to fetch from that series
s['cse']
```

```python
#unstack
temp = s.unstack() #for converting series to data frame
```

```python
#stack
temp.stack() #for converting data frame to series
```

Long vs wide data format

- wide format is where we have a single row for every data point with multiple columns to hold the values of various attributes

- long format is where, for each data point we have as many rows as the numbers of attributes and each  row cotains the value of a particular attribute for a given data point


```python
#melt -> simple example branch (wide to long data format)
pd.DataFrame({'cse':[120]}).melt()
```

```python
pd.DataFrame({'cse':[120],'ece':[100],'mech':[50]}).melt(var_name='branch',value_name='student')
```

```python
pd.DataFrame({
    'branch':['cse','ece','mech'],
    '2020':[100,150,60],
    '2021':[120,130,80],
    '2022':[150,140,70],
}).melt(id_vars=['branch'],var_name='year',value_name='students')
```

```python
#melt ->real world example
death = pd.read_csv("time_series_covid19_deaths_global.csv")
confirm = pd.read_csv("time_series_covid19_confirmed_global.csv")
```

```python
death = death.melt(id_vars=['Province/State','Country/Region','Lat','Long'],var_name='date',value_name='no.of deaths')
confirm = confirm.melt(id_vars=['Province/State','Country/Region','Lat','Long'],var_name='date',value_name='no.of cases')
```
