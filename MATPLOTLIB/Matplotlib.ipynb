```python
#import the library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

- Bivariate Analysis
- categorical -> numerical and numerical -> numerical
- use case - Time series data


```python
#plotting a simple function

price = [48000, 54000, 57000, 49000, 47000, 45000]
year = [2015, 2016, 2017, 2018, 2019, 2020]

plt.plot(year, price)
plt.xlabel("Year")
plt.ylabel("Price")
plt.title("Price Trend Over Years")
plt.show()

```

```python
batsman=pd.read_csv('sharma-kohli.csv')
batsman.set_index('index')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
#label title 
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit and virat career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
#for different colors 
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')

plt.title('Rohit and virat career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
#line style using like dasdot,dotted,simpleline , we can increase the linewidth
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dotted')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle="dotted")

plt.title('Rohit and virat career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',marker='D')

plt.title('Rohit and virat career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
#legend -> location of the points 
plt.plot(batsman['index'],batsman['V Kohli'],color='green',marker='D',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',marker='D',label='Rohit')

plt.title('Rohit and virat career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs')
plt.legend()
```

```python
#limit axes
price = [48000, 54000, 57000, 49000, 47000, 4500000]
year = [2015, 2016, 2017, 2018, 2019, 2020]

plt.plot(year, price)

```

for limit axes we can use plt.ylim(0,what ever limit you want)

Scatter plot
- Bivariate Analysis
- Numerical vs numerical
- use case - finding correlation

```python
#plt.scatter simple function 
x=np.linspace(-10,10,50)

y=10*x + 3+np.random.randint(0,300,50)
y
```

```python
plt.scatter(x,y)
```

```python
# plt .scatter plt on pandas data
batter=pd.read_csv('batter.csv')
first_fifty=batter.head(50)
first_fifty
```

```python
plt.scatter(first_fifty['avg'],first_fifty['strike_rate'])
```

```python
plt.scatter(first_fifty['avg'],first_fifty['strike_rate'])
plt.title("Avg and Strike rate of top 50 Batsman")
plt.xlabel('Average')
plt.ylabel('Strike_rate')
```

```python
#marker and color also we can easily use in this plt
plt.scatter(first_fifty['avg'],first_fifty['strike_rate'],marker='.')
plt.title("Avg and Strike rate of top 50 Batsman")
plt.xlabel('Average')
plt.ylabel('Strike_rate')
```

```python
#size
tips=sns.load_dataset('tips')
tips
```

```python
plt.scatter(tips['total_bill'],tips['tip'])
```

```python
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
```

```python
#scatter plot using plt.plot
#faster for huge data set
plt.plot(tips['total_bill'],tips['tip'],'o')
```

plt.plot
-Purpose: Draws continuous lines (or markers connected by lines).

-Usage: Best for functions, trends, or data that should be connected.

-Default behavior



plt.scatter
-Purpose: Plots individual points (no connecting lines).

-Usage: Best for showing distributions, relationships, or clusters.

-Default behavior: Displays unconnected dots.
