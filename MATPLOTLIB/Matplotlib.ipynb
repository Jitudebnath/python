plt.plot
-Purpose: Draws continuous lines (or markers connected by lines).

-Usage: Best for functions, trends, or data that should be connected.

-Default behavior



plt.scatter
-Purpose: Plots individual points (no connecting lines).

-Usage: Best for showing distributions, relationships, or clusters.

-Default behavior: Displays unconnected dots.

3D surface plots

```python
#scatter + surface together
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx,yy = np.meshgrid(x,y)
```

```python
z=xx**2+yy**2
z.shape
```

```python
fig=plt.figure(figsize=(12,8))

ax = plt .subplot(projection='3d')

p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
plt.show()
```

Contour Plots

```python
fig=plt.figure(figsize=(12,8))

ax = plt .subplot()

p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
plt.show
```

```python
fig=plt.figure(figsize=(12,8))

ax = plt .subplot()

p=ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
plt.show
```
