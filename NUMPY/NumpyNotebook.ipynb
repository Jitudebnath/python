```python
#Array slicing
arr = [10, 20, 30, 40, 50]

print(arr[1:4])     # [20, 30, 40]
print(arr[:3])      # [10, 20, 30]
print(arr[::2])     # [10, 30, 50]
print(arr[::-1])    # for ending to starting of a number
```

```python


# Create a 3D NumPy array
Arr = np.array([
    [[0, 1], [2, 3]],
    [[4, 5], [6, 7]]
])

print("Original 3D Array:\n", Arr)

# 🔹 Slice examples
print("\nFirst block:\n", Arr[0])          # [[0 1] [2 3]]
print("\nSecond block:\n", Arr[1])         # [[4 5] [6 7]]
print("\nFirst row of first block:", Arr[0, 0])   # [0 1]
print("\nSingle element [1,1,0]:", Arr[1, 1, 0])  # 6

# 🔹 Print all elements (flatten)
print("\nAll elements flattened:", Arr.flatten())
```
