```python
import numpy as np

# Create a 2D array
arr = np.arange(12).reshape(3,4)
print("Original Array:\n", arr)

# Fancy indexing: select specific rows and columns
rows = [0, 2]       # first and last row
cols = [1, 3]       # second and last column

# Extract elements using fancy indexing
result = arr[rows, cols]
print("\nFancy Indexed Result:", result)

# Another example: select multiple rows and columns
result2 = arr[[0,1,2], [3,2,1]]
print("\nFancy Indexed Result 2:", result2)
```

Broadcasting

The term broadcasting describes how Numpy treats arrays with different
shapes during arithmetic operations.
The smaller array is "broadcast" across the large array so that they have compatible shapes.

```python
arr1=np.arange(6).reshape(2,3)
arr2=np.arange(6,12).reshape(2,3)

print(arr1)
print(arr2)

print(arr1+arr2)
```

```python
arr1=np.arange(6).reshape(2,3)
arr2=np.arange(3).reshape(1,3)

print(arr1)
print(arr2)

print(arr1+arr2)
```
