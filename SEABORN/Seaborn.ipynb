2 . Distribution Plots

- used for univariate analysis
- used to find out the distribution
- Range of the obervation
- Central Tendency
- is the data bimodal?
- Are there outliers?

Plot under distribution plot
- histplot
- kdeplot
- rugplot


```python
#figure level function ->displot
#axes level function -> histplot -> kdeplot -> rugplot
```

```python
#plotting univariate histogram
sns.histplot(data=tips,x='total_bill')
```

```python
#displot
sns.displot(data=tips,kind='hist',x='total_bill')
plt.show()
```

```python
#it's also possible to visualize the distribution of categorical variable using the logic of a histogram.
# Discrete bins are automatically set for categorical variables 
tips
```

```python
sns.displot(data=tips,
            kind='hist',
            x='day')
plt.show()
            
```

```python
#hue parameter
sns.displot(data=tips,
            kind='hist',
            x='tip',
            hue='sex')
plt.show()
```

```python
#element using col and row
sns.displot(data=tips,
            kind='hist',
            x='tip',
            hue='sex',
            element='step')
plt.show()
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,
            kind='hist',
            x='age',
            element='step',
            hue='sex')
plt.show()
```

```python
#faceting using col and row
sns.displot(data=tips,
            kind='hist',
            x='tip',
            element='step',
            col='sex')
plt.show()
```

```python
#kdeplot
#Rather than using discrete bins, a kde plot smooths the observations with a gaussian kernel,producing a continous density estimate
sns.kdeplot(data=tips,
            x='total_bill')
plt.show()
```

```python
#hue->fill
sns.kdeplot(data=tips,
            x='total_bill',
            hue='sex',
            fill=True)
plt.show()
```

```python
#Rugplot

#plot marginal distribution by drawing ticks along x and y axes.

#This function is intended to complement other plot by showing the location of individual observations in an unobtrusive way.

sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
#Bivariate histogram
# a bivariate histogram bins the data within rectangle that tile the plot
# and then shows the count of observation within each reactangle with the fill color

sns.histplot(data=tips,
            x='total_bill',
            y='tip')
plt.show()
```

```python
#Bivariate kdeplot
# a bivariate kde plot smothes the (x,y) observations witha 2D gaussian
sns.kdeplot(data=tips,x='total_bill',y='tip')
plt.show()
```

2.Matrix plot
- Heatmap
- Clustermap

```python
#Heatmap
#plot rectangular data as a color-encoded matrix
gap
```

```python
temp=gap.pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(10,10))
sns.heatmap(temp,cmap='viridis') #axes level function 
```

```python
#annot
europe=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(10,10))
sns.heatmap(europe,annot=True,linewidths=0.5,cmap='viridis')
plt.show()
```

```python
#clustermap

#plot a matrix dataset as a hierarchically-clustered heatmap .
#This function reqires scipy to be available

px.data.iris()
```

```python
iris=sns.load_dataset('iris')
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]],cmap='viridis')
plt.show()
```

Figure level function -> catplot


```python
#Strip plot
sns.stripplot(data=tips,x='day',y='total_bill')#use jitter for cancel noise 
plt.show()
```

```python
#using catplot , we can control jitter value here
sns.catplot(data=tips,x='day',y='total_bill',kind='strip',jitter=0.2,hue='sex')
plt.show()
```

```python
#swarmplot
sns.catplot(data=tips,x='day',y='total_bill',kind='swarm',hue='sex')
plt.show()
```

BoxPlot

- A box plot (also called a box and whisker plot) is a standardized visual graph that displays the distribution, variation, and outliers of numerical data

```python
# Box plot
sns.boxplot(data=tips,x='day',y='total_bill')
plt.show()
```

```python
#using catplot
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
plt.show()
```

```python
#hue
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
sns.boxplot(data=tips,y='total_bill')
```

Violin Plot = (boxplot+kdeplot)

```python
#violinplot
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
sns.violinplot(data=tips,x='day',y='total_bill',hue='sex')
plt.show()
```

Barplot

A barplot (or bar chart) is a graphical display of data using rectangular bars where the length or height of the bar is proportional to the value it represents


```python
sns.barplot(data=tips,x='sex',y='total_bill',hue='smoker')
plt.show()
```

when there are multiple observation in each category , it also use bootstraping to compute a confidence interval around the estimate ,which  is plotted using error bars

```python
sns.barplot(data=tips,x='sex',y='total_bill',hue='smoker',ci=None)
plt.show()
```

```python
#pointplot
sns.pointplot(data=tips,x='sex',y='total_bill',hue='smoker')
plt.show()
```

```python
sns.pointplot(data=tips,x='sex',y='total_bill',hue='smoker',ci=None)
plt.show()
```

CountPlot

A countplot is a categorical bar chart that visually displays the frequency (or count) of observations in each category

```python
#Count plot use for categorical value
sns.countplot(data=tips,x='sex',hue='day')
plt.show()

```

```python
#faceting using catplot
sns.catplot(data=tips,x='sex',y='total_bill',col='smoker',kind='box',row='time')
plt.show()
```

Regression Plots
- regplot
- Implot

In the simplest invocation,both function draw a scatterplot of two variables,x and y, and then fit the regression model y~x and plot the resulting regrression line and a 95% confidence for that regression

```python
#regplot is a axes level function 
sns.regplot(data=tips,x='total_bill',y='tip')
plt.show()
```

```python
#lmplot
sns.lmplot(data=tips,x='total_bill',y='tip')
plt.show()
```

```python
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
plt.show()
```

```python
#residplot
sns.residplot(data=tips,x='total_bill',y='tip')
plt.show()
```
