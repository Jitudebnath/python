```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px 
```

Main Classification

```python
tips=sns.load_dataset('tips')
tips
```

```python
#scatter plot -> axes level function
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter')
```

```python
#using hue
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex')
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="scatter",
    hue="sex",
    style="smoker",
    size="total_bill"
)
```

```python
#line plot
gap = px.data.gapminder()
gap
```

```python
temp_df=gap[gap['country']=='India']
temp_df
```

```python
sns.lineplot(data=temp_df,x='year',y='lifeExp')
```

```python
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line')
```

```python
temp_df=gap[gap['country'].isin(['India','Pakistan','China'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country')
```

```python
#facet Plot
# using row and cols we can use this facet plot and it works with figure level function
#it will not work with scatterplot and lineplot
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='sex',row='smoker')
```

```python
#line plot
sns.relplot(data=tips,
            x='total_bill',
            y='tip',
            kind='line',
            col='sex',
            row='smoker')
```

```python
#col wrap 
sns.relplot(
    data=gap,
    x='lifeExp',
    y='gdpPercap',
    kind='scatter',
    col='year',
    col_wrap=4
)
```
