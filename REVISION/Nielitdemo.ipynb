```python
#palindrome check
word = 'madam'
is_palindrome = True
l=len(word)

for i in range (l//2):
    if word[i] != word[l-1-i]:
        is_palindrome = False 
        break

if is_palindrome:
    print('palindrome')
else:
    print('Not a palindrome')    
```

```python
word = 'madam'
resv = word [::-1]
if word==resv:
    print(f'{word} -> palindrome')
else:
    print(f'{word} -> not a palindrome')    
```

```python
#write a program to reversing a number
num=6789
rev_num = 0

print(f"original number = ",num)

for _ in range(100):
    rem = num % 10
    rev_num = rev_num * 10 + rem
    num //= 10 
    if num == 0 :
        break

print("Reversed number = ", rev_num)
```

```python
#counting the frequency of elements in a list using a dictionary
list_val = [1,2,3,3,4,5,7,8,5,6,7,8]
print(list_val)
print()
freq = {}
for item in list_val:
    if (item in freq):
        freq[item]+-1
    else:
        freq[item] = 1

for key , value in freq.items():
    print("%d : %d" %(key,value))            
```

```python
#Recursio using python programming language

def factorical(n):
    if n==1 or n==0:
        return 1
    else:
        return(n*factorical(n-1))
    
num=5
print("factorial:",factorical(num))    
```

```python
#sumation of n number using recursion

def sumation(n):
    if n==0:
        return 0
    else:
        return (n+sumation(n-1))

num=5
print("sumation:",sumation(num))
```
