1.Scatterplot(Numerical-Numerical)

```python
tips.plot(kind='scatter',x='total_bill',y='tip')
plt.show()
```

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'])
plt.show()
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",       # color by gender
    style="smoker"   # marker style by smoker status
)
plt.show()
```

```python
sns.boxplot(
    data=titanic,
    x='Sex',
    y='Age',
    hue='Survived'
)
plt.show()
```

4.Distplot(numberical-categorical)

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'],hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'],hist=False)
plt.show()
```

5.Heatmap (categorical-categorical)

```python
titanic.head(5)
```

```python
pd.crosstab(titanic['Pclass'],titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
titanic.groupby("Embarked")["Survived"].mean().mul(100).round(2)
```

6. clustermap

```python
sns.clustermap(pd.crosstab(titanic['SibSp'],titanic['Survived']))
```

8.Line plot(Numerical-Numerical)

```python
sns.load_dataset('flights')
```

```python
new=flights.groupby('year').sum(numeric_only=True).reset_index()
```

```python
sns.lineplot(x=new['year'],y=new['passengers'])
```

```python
flights.pivot_table(values='passengers',index='month',columns='year')
```

```python
sns.heatmap(flights.pivot_table(values='passengers',index='month',columns='year'),cmap='viridis')
plt.show()
```

```python
sns.clustermap(flights.pivot_table(values='passengers',index='month',columns='year'),cmap='viridis')
```
