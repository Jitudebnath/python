```python
import numpy as np 
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic=pd.read_csv('Titanic.csv')
titanic
```

Univariate analysis


a.Countplot

```python
sns.countplot(titanic['Survived'])
titanic['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(titanic['Pclass'])
```

```python
sns.countplot(titanic['Sex'])
```

b.Pie Chart

```python
titanic['Survived'].value_counts().plot(kind='pie',autopct='%.2f')
```

```python
titanic['Pclass'].value_counts().plot(kind='pie',autopct='%.2f')
```

```python
titanic['Sex'].value_counts().plot(kind='pie',autopct='%.2f')
```

1.Scatterplot(Numerical-Numerical)

```python
tips.plot(kind='scatter',x='total_bill',y='tip')
plt.show()
```

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'])
plt.show()
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",       # color by gender
    style="smoker"   # marker style by smoker status
)
plt.show()
```
